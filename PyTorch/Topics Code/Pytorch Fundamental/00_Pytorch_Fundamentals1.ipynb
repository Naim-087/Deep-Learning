# Creating Tensor

A Tensor is a multidimensional array used to store and process numerical data in PyTorch.

| Concept   | Example                             |
| --------- | ----------------------------------- |
| Scalar    | `5`                                 |
| Vector    | `[1, 2, 3]`                         |
| Matrix    | `[[1,2], [3,4]]`                    |
| 3D Tensor | Multiple matrices                   |
| 4D Tensor | Commonly used for batches of images |

## Random Tensor

> A random tensor is a tensor whose values are generated randomly according to a specified distribution.

> In PyTorch, random tensors are commonly used for weight initialization, testing, simulations, and creating sample data.

**0.Tensor Range**

```python
number = torch.arange(start = 0 ,end=1000,step = 80)
number
```

**1. Random values between 0 and 1**

```python
x = torch.rand(3,4)         #torch.rand(row,col)
print(x)
```

```python
x.ndim
```

**2. Random Integer**

```python
x = torch.randint(0,10,(3,4))       #torch.randint(range,(shape))
x
```

```python
x.shape
```

**3. Random numbers from normal distribution**

```python
x = torch.randn(4,3)

print(x)
```

| Function          | Values              | Distribution    |
| ----------------- | ------------------- | --------------- |
| `torch.rand()`    | 0 → 1               | Uniform         |
| `torch.randn()`   | Positive & negative | Normal          |
| `torch.randint()` | Integers            | Discrete random |

# Tensor Dtypes

 > A tensor data type (dtype) determines what kind of values a tensor stores and how much memory those values use.

```python


x = torch.tensor([1, 2, 3])

print(x.dtype)
```

```python
x = torch.tensor([1, 2, 3], dtype=torch.int32)
```

## Integer types

| dtype         | Description    |
| ------------- | -------------- |
| `torch.int8`  | 8-bit integer  |
| `torch.int16` | 16-bit integer |
| `torch.int32` | 32-bit integer |
| `torch.int64` | 64-bit integer |


## Floating-point types

```python
x = torch.tensor([1.0, 2.0, 3.0])
print(x.dtype)
```

| dtype     |   Size | Typical use                            |
| --------- | -----: | -------------------------------------- |
| `float16` | 16-bit | Faster/memory-efficient training       |
| `float32` | 32-bit | **Default/common deep learning dtype** |
| `float64` | 64-bit | High numerical precision               |


```python
x = torch.tensor([1.5, 2.5], dtype=torch.float32)
x.dtype
```

## Boolean

```python
x = torch.tensor([True, False, True])

print(x.dtype)
```

## Complex numbers

```python
x = torch.tensor([1+2j, 3+4j])

print(x.dtype)
```
