# PyTorch: Maximum, Mean, and Sum

This Colab notebook covers:

- `torch.max()`
- `torch.mean()`
- `torch.sum()`
- Finding values in a specific column
- Finding values in a specific row
- Understanding `dim=0` and `dim=1`
- Getting both values and indices with `torch.max()`
- Several practical examples


````python
# Import PyTorch
import torch

print("PyTorch version:", torch.__version__)
````

## 1. Create a 2D Tensor

We will use this tensor for the first few examples:

```text
30  31  40  35
27  35  30  40
32  40  35  25
35  38  25  30
```

Shape = `(4, 4)`

- 4 rows
- 4 columns


````python
X = torch.tensor([
    [30, 31, 40, 35],
    [27, 35, 30, 40],
    [32, 40, 35, 25],
    [35, 38, 25, 30]
])

print("X:")
print(X)

print("\nShape:", X.shape)
````

# 2. `torch.max()`

`torch.max()` finds the maximum value.

## Maximum of the entire tensor

```python
torch.max(X)
```


````python
max_value = torch.max(X)

print("Maximum value:", max_value)
````

## Maximum of each column

Use `dim=0`.

```python
torch.max(X, dim=0)
```

`dim=0` means operate **down the rows**, giving one result for each column.


````python
max_values, max_indices = torch.max(X, dim=0)

print("Maximum of each column:")
print(max_values)

print("\nRow index of each maximum:")
print(max_indices)
````

### Understanding the result

For the tensor above:

- Column 0 → maximum = 35, row index = 3
- Column 1 → maximum = 40, row index = 2
- Column 2 → maximum = 40, row index = 0
- Column 3 → maximum = 40, row index = 1


## Maximum of each row

Use `dim=1`.

```python
torch.max(X, dim=1)
```


````python
max_row_values, max_row_indices = torch.max(X, dim=1)

print("Maximum of each row:")
print(max_row_values)

print("\nColumn index of each maximum:")
print(max_row_indices)
````

## Maximum in a specific column

Suppose we want **column 2**.

```python
X[:, 2]
```

means:

> Take every row from column 2.

Then apply `torch.max()`.


````python
column_number = 2

max_value, max_index = torch.max(X[:, column_number], dim=0)

print("Column:", column_number)
print("Column values:", X[:, column_number])
print("Maximum value:", max_value)
print("Row index:", max_index)
````

## Maximum in a specific row

Suppose we want **row 1**.

```python
X[1, :]
```

means:

> Take every column from row 1.


````python
row_number = 1

max_value, max_index = torch.max(X[row_number, :], dim=0)

print("Row:", row_number)
print("Row values:", X[row_number, :])
print("Maximum value:", max_value)
print("Column index:", max_index)
````

# 3. `torch.mean()`

`torch.mean()` calculates the arithmetic average.

Formula:

**mean = sum of values / number of values**


````python
# Mean of the entire tensor

mean_value = torch.mean(X.float())

print("Mean of entire tensor:", mean_value)
````

### Why `.float()`?

Our tensor contains integers. `torch.mean()` generally requires a floating-point tensor, so we use:

```python
X.float()
```


````python
# Mean of each column

column_means = torch.mean(X.float(), dim=0)

print("Mean of each column:")
print(column_means)
````

````python
# Mean of each row

row_means = torch.mean(X.float(), dim=1)

print("Mean of each row:")
print(row_means)
````

````python
# Mean of a specific column

column_number = 1

column_mean = torch.mean(X[:, column_number].float())

print("Column values:", X[:, column_number])
print("Mean:", column_mean)
````

````python
# Mean of a specific row

row_number = 2

row_mean = torch.mean(X[row_number, :].float())

print("Row values:", X[row_number, :])
print("Mean:", row_mean)
````

# 4. `torch.sum()`

`torch.sum()` adds values together.

## Sum of the entire tensor


````python
total_sum = torch.sum(X)

print("Sum of entire tensor:", total_sum)
````

````python
# Sum of each column

column_sums = torch.sum(X, dim=0)

print("Sum of each column:")
print(column_sums)
````

````python
# Sum of each row

row_sums = torch.sum(X, dim=1)

print("Sum of each row:")
print(row_sums)
````

````python
# Sum of a specific column

column_number = 3

column_sum = torch.sum(X[:, column_number])

print("Column values:", X[:, column_number])
print("Sum:", column_sum)
````

````python
# Sum of a specific row

row_number = 0

row_sum = torch.sum(X[row_number, :])

print("Row values:", X[row_number, :])
print("Sum:", row_sum)
````

# 5. Different Example: Student Marks

Suppose each row represents a student and each column represents a subject.

```text
Math  Physics  Programming
85      78        92
72      88        80
95      91        89
60      75        70
```


````python
marks = torch.tensor([
    [85, 78, 92],
    [72, 88, 80],
    [95, 91, 89],
    [60, 75, 70]
])

print(marks)
````

````python
# Highest mark in each subject

subject_max = torch.max(marks, dim=0)

print("Highest mark in each subject:")
print(subject_max.values)

print("\nStudent index for each highest mark:")
print(subject_max.indices)
````

````python
# Average mark in each subject

subject_mean = torch.mean(marks.float(), dim=0)

print("Average mark in each subject:")
print(subject_mean)
````

````python
# Total marks obtained by each student

student_total = torch.sum(marks, dim=1)

print("Total marks for each student:")
print(student_total)
````

````python
# Average mark for each student

student_average = torch.mean(marks.float(), dim=1)

print("Average mark for each student:")
print(student_average)
````

# 6. Different Example: Sensor Data

Imagine four measurements from three sensors:

- Column 0 → Heart Rate
- Column 1 → Temperature
- Column 2 → Oxygen Level

This type of operation is useful when working with health or sensor datasets.


````python
sensor_data = torch.tensor([
    [72, 36, 98],
    [80, 37, 97],
    [75, 36, 99],
    [90, 38, 95]
])

print(sensor_data)
````

````python
# Maximum value of each sensor

max_sensor = torch.max(sensor_data, dim=0)

print("Maximum:")
print(max_sensor.values)
print("\nIndices:")
print(max_sensor.indices)
````

````python
# Mean value of each sensor

mean_sensor = torch.mean(sensor_data.float(), dim=0)

print("Mean:")
print(mean_sensor)
````

````python
# Total of each sensor

sum_sensor = torch.sum(sensor_data, dim=0)

print("Sum:")
print(sum_sensor)
````

# 7. Quick Comparison

| Operation | Entire Tensor | Each Column | Each Row |
|---|---|---|---|
| Maximum | `torch.max(X)` | `torch.max(X, dim=0)` | `torch.max(X, dim=1)` |
| Mean | `torch.mean(X.float())` | `torch.mean(X.float(), dim=0)` | `torch.mean(X.float(), dim=1)` |
| Sum | `torch.sum(X)` | `torch.sum(X, dim=0)` | `torch.sum(X, dim=1)` |

### Remember

```text
dim=0 → column-wise result
dim=1 → row-wise result
```

For a specific column:

```python
X[:, column_number]
```

For a specific row:

```python
X[row_number, :]
```
