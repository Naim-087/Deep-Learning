## 1. Tensor Shape

The `shape` of a tensor tells us how many elements it has along each dimension.

Example:

```text
[[1, 2, 3],
 [4, 5, 6]]

````python

X = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(X)
print("Shape:", X.shape)
print("Number of dimensions:", X.ndim)
````

## Reshaping

Reshaping changes the shape of a tensor while keeping the same elements.

For example:

```text
Original: 2 × 3

[[1, 2, 3],
 [4, 5, 6]]

Reshape: 3 × 2

[[1, 2],
 [3, 4],
 [5, 6]]

````python

X = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original:")
print(X)

print("\nOriginal shape:")
print(X.shape)

Y = X.reshape(3, 2)

print("\nReshaped:")
print(Y)

print("\nNew shape:")
print(Y.shape)
````

## View

view() is another common way to change the shape of a tensor.

```python
X.view(3, 4)

````python

## 🟦 Cell 11


X = torch.arange(12)

Y = X.view(3, 4)

print("Original:")
print(X)

print("\nViewed as 3 x 4:")
print(Y)
````

## Flatten



Flatten converts a tensor into a single dimension.

Example:

```text
2 × 3

[[1, 2, 3],
 [4, 5, 6]]

↓

[1, 2, 3, 4, 5, 6]

````python


X = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

Y = torch.flatten(X)

print("Original:")
print(X)

print("\nFlattened:")
print(Y)

print("\nShape:")
print(Y.shape)
````

## Stacking

`torch.stack()` combines tensors by creating a **new dimension**.

Suppose:

```text
A = [1, 2, 3]

B = [4, 5, 6]

````python


A = torch.tensor([1, 2, 3])
B = torch.tensor([4, 5, 6])

C = torch.stack([A, B])

print("A:")
print(A)

print("\nB:")
print(B)

print("\nStacked:")
print(C)

print("\nShape:", C.shape)
````

````python
A = torch.tensor([1, 2, 3])
B = torch.tensor([4, 5, 6])
C = torch.tensor([7, 8, 9])

result = torch.stack([A, B, C])

print(result)
print("Shape:", result.shape)
````

##  Concatenation

`torch.cat()` joins tensors along an **existing dimension**.

This is different from `stack()`.

### `stack()`

Creates a new dimension.

### `cat()`

Uses an existing dimension.

````python
A = torch.tensor([
    [1, 2],
    [3, 4]
])

B = torch.tensor([
    [5, 6],
    [7, 8]
])

print("A:")
print(A)

print("\nB:")
print(B)

print("\nConcatenate along dim=0:")
print(torch.cat([A, B], dim=0))

print("\nConcatenate along dim=1:")
print(torch.cat([A, B], dim=1))
````

## Unsqueeze


---







`unsqueeze()` adds a dimension of size `1`.

Example:

```text
Original:

[1, 2, 3]

Shape:
(3)

↓

unsqueeze(0)

[[1, 2, 3]]

Shape:
(1, 3)

````python

X = torch.tensor([1, 2, 3])

print("Original:")
print(X)

print("Shape:", X.shape)

Y = X.unsqueeze(0)

print("\nAfter unsqueeze(0):")
print(Y)

print("Shape:", Y.shape)
````

## Squeeze


`squeeze()` removes dimensions whose size is `1`.

Example:

```text
[[1, 2, 3]]

Shape:
(1, 3)

↓

squeeze()

[1, 2, 3]

Shape:
(3)

````python

X = torch.tensor([[1, 2, 3]])

print("Original:")
print(X)

print("Shape:", X.shape)

Y = X.squeeze()

print("\nAfter squeeze:")
print(Y)

print("Shape:", Y.shape)
````

## Unsqueeze → Squeeze

````python
X = torch.tensor([1, 2, 3])

print("Original:")
print(X)
print("Shape:", X.shape)

Y = X.unsqueeze(0)

print("\nUnsqueeze:")
print(Y)
print("Shape:", Y.shape)

Z = Y.squeeze(0)

print("\nSqueeze:")
print(Z)
print("Shape:", Z.shape)
````
